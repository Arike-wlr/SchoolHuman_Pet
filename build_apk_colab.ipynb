```python
# 安装依赖
!pip install buildozer cython
!apt-get update && apt-get install -y git zip unzip openjdk-17-jdk python3-pip autoconf libtool pkg-config zlib1g-dev libncurses5-dev libncursesw5-dev libtinfo5 cmake libffi-dev libssl-dev
```

```python
# 克隆项目（请替换为你的仓库地址）
!git clone https://github.com/yourusername/SchoolHuman_Pet.git
cd SchoolHuman_Pet
```

```python
# 创建 buildozer.spec 配置
%%writefile buildozer.spec
[app]

title = SchoolHumanPet
package.name = schoolhumanpet
package.domain = org.example
source.dir = .
source.include_exts = py,png,jpg,kv,atlas,json,txt
version = 0.1.0

requirements = python3,kivy,requests,markdown,websocket-client
android.add_assets = pet/,customized/,高校拟人OC_1位角色_2026-07-13.json,南京高校_2026-07-13.json,userinfo.txt

android.api = 33
android.ndk = 25b
android.sdk = 24
android.buildtools = 33.0.0

android.permissions = INTERNET
android.icon = %(source.dir)s/pet/pet.png

[buildozer]
log_level = 2
warn_on_root = 1
```

```python
# 构建 APK（首次构建耗时较长，约15-30分钟）
!buildozer android debug
```

```python
# 下载 APK
from google.colab import files
files.download('bin/SchoolHumanPet-0.1.0-debug.apk')
```
